# 1 — Jupyter Notebooks

## 1.1 What is a Jupyter Notebook?

A **Jupyter notebook** is a document that mixes:

- **Text** (explanations, headings, images, formulas) — written in *Markdown*
- **Code** (small blocks of Python you can run)
- **Output** (results, tables, plots) — shown directly below the code that produced it

A notebook file has the extension **`.ipynb`** (short for *IPython notebook*).

You can think of a notebook like a **lab journal**: a sequence of blocks where you write down what you are doing, run a small experiment, and see the result — all in one place.

## 1.2 Cells — the building blocks

A notebook is made of **cells**. Each cell is a small block that can be one of two types:

| Cell type | What it contains | What happens when you run it |
|---|---|---|
| **Markdown cell** | Formatted text (like this one) | The text is rendered nicely |
| **Code cell** | Python code | The code runs, the output appears below |

## 1.3 How do I run a cell?

There are three common ways to run the currently selected cell:

- **`Shift` + `Enter`** → run the cell **and move to the next one** (most common)
- **`Ctrl` + `Enter`** → run the cell **and stay on it**
- **Click the little ▶ (play) button** at the top of the page

> Try it now: click on the cell above this one, press `Shift`+`Enter`, and watch the selection jump down.

## 1.4 Two modes: *edit* vs. *command*

When you click on a cell you are in **edit mode** (you can type inside it). The cell has a coloured border.

If you press **`Esc`**, you leave the cell and enter **command mode**. Now the keyboard controls the notebook itself. A few useful shortcuts in command mode:

| Key | Action |
|---|---|
| `A` | **A**dd a new cell **above** |
| `B` | Add a new cell **b**elow |
| `D`, `D` (twice) | **D**elete the current cell |
| `M` | Change cell to **M**arkdown |
| `Y` | Change cell to code (P**y**thon) |
| `Enter` | Go back to edit mode |

Don't worry about memorising these — knowing they exist is enough for today.

# 2 — Script vs. Notebook

## 2.1 What is a script?

A **Python script** is just a plain text file with the extension **`.py`**.
It contains Python code from top to bottom, and nothing else — no formatted text, no output stored inside.

Example — a file called `hello_world.py` might contain only:

```python
print("Hello")
print("World")
```

To **run** a script, you type a command in a terminal, for example:

```
uv hello_world.py
```
or
```
python hello_world.py
```

Python then reads the file **from top to bottom**, runs every line, prints the output to the terminal, and quits.

## 2.2 What is a notebook?

A **notebook** (`.ipynb`) is not run all at once. It is run **cell by cell**, in whatever order you choose.

Behind the notebook there is a small program called the **kernel** — think of it as "the Python that is listening to your notebook". When you run a cell:

1. The code is sent to the kernel.
2. The kernel runs it.
3. The result comes back and appears below the cell.
4. **The kernel remembers everything** you have run so far, until you restart it.

This is why notebooks feel *interactive*: you can try one small piece, look at the result, then try another piece.

## 2.3 The kernel remembers — a small demo

Run the next three code cells **in order** (`Shift`+`Enter` three times).

````python
x = 10
````

````python
y = 5
````

````python
x + y
````

The third cell knows about `x` and `y` — even though they were defined in earlier cells — because the kernel kept them in memory.

A **script** does not work like that: every time you run `python hello.py`, Python starts fresh, runs everything top-to-bottom, and forgets everything at the end.

## 2.4 Careful — cell order matters!

Because you can run cells **in any order**, it is easy to fool yourself. Look at the numbers in the `[ ]` on the left of each cell — they show the *actual* order in which cells were run, not the order in which they appear on screen.

Good habit: from time to time click **"Restart Kernel..."** (or **"Restart Kernel and Run All Cells..."**) to make sure your notebook still works from top to bottom, exactly like a script would.

## 2.5 When to use which?

| Notebook (`.ipynb`) | Script (`.py`) |
|---|---|
| Learning, exploring, experimenting | Automating a task, production code |
| Data analysis, plots, reports | Programs that must run without a human |
| Mix of text + code + results in one document | Pure code, easier to reuse and share as a program |
| Run cell by cell, keep intermediate results | Run once, top to bottom |
| Great for teaching and prototyping | Great for building real applications |

Both are just **text files with Python inside** — you will use both in your career.

# 3 — File Paths

## 3.1 Why do we care about paths?

Almost every real program has to **open a file**: a spreadsheet with data, an image, a configuration file, a log to write into…

To open a file, Python needs to know **where** the file is on your computer. That location is called a **path**.

There are two ways to describe *where* a file is:

- an **absolute path** — the complete address, from the very top of the disk
- a **relative path** — the address *starting from where I am right now*

Both refer to the same file — they are just two different ways to describe the road to it.

## 3.2 A tiny example folder

Imagine your computer contains this small folder tree:

```
C:\
└── Users\
    └── anna\
        └── course\
            ├── notebooks\
            │   └── lecture1.ipynb   ← we are working here
            └── data\
                └── students.csv     ← we want to open this file
```

(On macOS or Linux it would look almost the same, but the top would be `/` instead of `C:\`, and the separators would be `/` instead of `\`.)

The folder `course\` is a child of the folder `anna\` and a parent of the folders `notebooks\` and `data\`. These are on the samle level and siblings.

## 3.3 Absolute paths — the full address

An **absolute path** starts at the very top of the disk and lists **every folder** on the way to the file.

For our example, the absolute path of `students.csv` is:

- On Windows: `C:\Users\anna\course\data\students.csv`
- On macOS / Linux: `/Users/anna/course/data/students.csv`

Think of it like a **postal address** with country, city, street and house number: no matter where in the world you send the letter from, it will always arrive at the same place.

**Good news:** an absolute path always works, from anywhere.

**Bad news:** it only works on *this* computer. On someone else's computer the user is not `anna`, so the path is different. That makes absolute paths a poor choice when you want to share your code.

### A note about the two slashes

- Windows uses **backslashes**: `C:\Users\anna`
- macOS and Linux use **forward slashes**: `/Users/anna`

In Python, the backslash `\` has a special meaning inside text, so you will often see one of these three forms for a Windows path:

```python
"C:\\Users\\anna\\course\\data\\students.csv"   # double every backslash
r"C:\Users\anna\course\data\students.csv"        # "raw" string, put an r in front
"C:/Users/anna/course/data/students.csv"          # forward slashes also work on Windows in Python
```

We will not use these yet — just be aware that all three describe the *same* file.

## 3.4 The "current working directory"

When Python runs, it is always "standing" in some folder. That folder is called the **current working directory** (often shortened to *cwd*).

For a notebook, the current working directory is usually the folder that contains the `.ipynb` file.

Let's peek at it. Run the cell below.

````python
import os
os.getcwd()
````

The output above is *where this notebook thinks it is standing right now*. All relative paths will be interpreted starting from there.

## 3.5 Relative paths — starting from where I am

A **relative path** does **not** start at the top of the disk. It starts at the current working directory.

Going back to our example folder:

```
course\
├── notebooks\
│   └── lecture1.ipynb   ← current working directory
└── data\
    └── students.csv     ← we want this
```

From `notebooks\`, how do we get to `students.csv`?

1. Go **up** one folder (from `notebooks\` back to `course\`) → `..`
2. Go **into** `data\`
3. Take the file `students.csv`

So the relative path is:

```
..\data\students.csv        (Windows)
../data/students.csv        (macOS / Linux, also works in Python on Windows)
```

Two special names appear in relative paths:

| Symbol | Meaning |
|---|---|
| `.` | "the current folder" |
| `..` | "the folder one level up (the parent)" |

So `./notes.txt` means "the file `notes.txt` next to me", and `../notes.txt` means "the file `notes.txt` in my parent folder".

## 3.6 Absolute vs. relative — side by side

| | Absolute path | Relative path |
|---|---|---|
| Starts at | Top of the disk (`C:\` or `/`) | Current working directory |
| Example | `C:\Users\anna\course\data\students.csv` | `..\data\students.csv` |
| Depends on where I run the code? | No | **Yes** |
| Works on someone else's computer? | Usually not | **Yes**, if they keep the same folder structure |
| When to use | Rarely — mostly for very specific system files | Most of the time, especially in projects you share |

**Rule of thumb:** in your own projects, keep your data *inside* the project folder and use **relative paths**. Your code will then run on any computer where someone downloads your project.

## 3.8 Common mistake — the path that "suddenly" stops working

Imagine you write in your notebook:

```python
open("students.csv")
```

This is a relative path. It means: *"find `students.csv` in the current working directory"*.

It works today because your notebook and the file are in the same folder. Tomorrow you move the notebook into a subfolder → the file is not "next to" the notebook anymore → **error**.

That is why we take a moment to think about *from where* the path is being read. Later in the course, when you meet the `pathlib` library, you will get nicer tools to work with paths — but the ideas of **absolute** and **relative** will stay exactly the same.
