## 3 — The four everyday types

Python figures out a value's **type** from how you write it. The four you
meet on day one: `int` (whole number), `float` (decimal), `str` (text),
`bool` (`True`/`False`). Ask any value its type with `type()`.

````python
integer_type = 316
print(integer_type, "is of type", type(integer_type))
````

````python
float_type = 316.0
print(float_type, "is of type", type(float_type))
````

````python
string_type = "Hello World"
print(string_type, "is of type", type(string_type))
````

````python
bool_type = True
print(bool_type, "is of type", type(bool_type))
````

A classic beginner bug: gluing text and a number with `+`. The first
cell breaks on purpose; the second fixes it by converting the number to
text with `str()`. Note `str()` makes a new text value — it doesn't change
the original number.

````python
case_number = 316
message = "Case number: " + case_number
````

````python
case_number = 316
message = "Case number: " + str(case_number)
print(message)
````

Two ways to turn a decimal into a whole number, and they differ:
`int()` **truncates** (chops the decimal off), `round()` **rounds**. So
`int(5.9)` is `5`, but `round(5.9)` is `6`.

````python
i = int(5.9)
i
````

````python
r = round(5.9)
r
````

````python
type(r)
````

Converting the other way: text to a number, number to text.

````python
f = float("3.14")
f
````

````python
h = float("Hello")
h
````

````python
s = str(2024)
s
````

Comparisons ask a yes/no question and return a `bool`. Note the
**double** `==` for "is equal to" — a single `=` would try to assign.
Confusing `=` and `==` is the number-one beginner bug.

````python
tariff_rate = 5.5

result = tariff_rate > 10
print(result)
````

````python
result = tariff_rate < 10
print(result)
````

````python
result = tariff_rate == 10
print(result)
````

````python
result = tariff_rate != 10
print(result)
````

Combine conditions with `and` / `or` — plain English words, not
symbols. `and` needs both sides true; `or` needs at least one.

````python
tariff_rate = 5.5

result = tariff_rate > 10 and tariff_rate < 20
print(result)
````

````python
result = tariff_rate > 10 or tariff_rate < 20
print(result)
````

## 4 — String manipulations

Strings come with built-in helper **methods**. Each one returns a **new**
string; the original never changes. We'll test them on a deliberately
messy filename.

````python
filename = "  Costa Rica - Tariff Schedule FINAL.csv  "
filename
````

````python
filename.strip()
````

````python
filename.lower()
````

````python
filename.strip().lower()
````

````python
filename.replace(" ", "_")
````

````python
filename.endswith(".csv")
````

````python
filename.strip().endswith(".csv")
````

````python
len(filename)
````

Notice `filename` is still the messy original — methods don't change
it. To keep a cleaned version, you must **assign** the result to a
variable.

````python
filename
````

````python
cleaned = filename.strip().lower().replace(" ", "_")
cleaned
````

You can reach into a string by position. Counting starts at **0**.
A slice `[a:b]` runs from `a` up to *but not including* `b`. `-1` means
the last character.

````python
cleaned[0]
````

````python
cleaned[5]
````

````python
cleaned[3:6]
````

````python
cleaned[-1]
````

Strings are **immutable** — you can't change one character in place.
The first cell tries and fails; read the `TypeError`. The workaround is to
build a new string with `.replace()`.

````python
cleaned[3] = 'X'
cleaned
````

````python
var = cleaned[3]
var
````

````python
replaced = cleaned.replace(var, "X")
replaced
````

````python
replaced2 = cleaned.replace(var, "X", 1)
replaced2
````

## 5 — f-strings

Building a sentence with variables inside is cleanest with an
**f-string**: put an `f` before the quotes and drop variable names inside
`{ }`. Compare the three approaches below — the f-string wins.

````python
country = "Kenya"
tariff_rate = 5.5
year = 2024
````

````python
print("In", year, ",", country, "had a tariff rate of", tariff_rate, "%", ".")
````

````python
print("In " + str(year) + ", " + country + " had a tariff rate of " + str(tariff_rate) + "%.")
````

````python
print(f"In {year}, {country} had a tariff rate of {tariff_rate}%.")
````

f-strings can also format numbers — for example, fix the number of
decimal places with `:.2f`.

````python
tariff_rate = 5.12986

print(f"{tariff_rate}")
print(f"{tariff_rate:.2f}")
print(f"{tariff_rate:.4f}")
````

## 6 — Lists

A **list** is an ordered collection, written with square brackets. It can
hold anything, and `len()` tells you how many items.

````python
my_list = [21, 0.512, "Hello", True]
print(my_list)
print(len(my_list))
````

Indexing and slicing work exactly like strings: position 0 is first,
`[a:b]` is half-open, leaving a side blank means "to the end" or "from the
start".

````python
my_list[1]
````

````python
my_list[0:3]
````

````python
my_list[:3]
````

````python
my_list[2:]
````

````python
my_short_list = my_list[1:4]
print(my_list)
print(my_short_list)
````

Unlike strings, lists **can** be changed in place. Watch each
operation modify `my_list`:

- `append` adds to the end,
- `pop(3)` removes by **index**,
- `[1] = ...` replaces by index,
- `remove("France")` removes by **value**,
- `insert(0, ...)` adds at a position.

````python
my_list.append("Switzerland")
my_list
````

````python
my_list.pop(3)
my_list
````

````python
my_list[1] = "France"
my_list
````

````python
my_list.remove("France")
my_list
````

````python
my_list.insert(0, "Germany")
my_list
````

## 8 — Dictionaries

A **dictionary** looks things up by **name** (a key) instead of by
position. You write it with `key: value` pairs in curly braces.

````python
country = {"name" : "Kenya", "population_millions" : 55.1}
country
````

````python
country["name"]
````

Add a key just by assigning to it. Add several at once with `.update()`.

````python
country["capital"] = "Nairobi"
country
````

````python
country.update({"region": "Africa", "wto_member": True})
country
````

The shape real data usually takes: a **list of dictionaries**, one
dict per record. To reach a field, index twice — first the record, then
the key.

````python
country2 = {
 'name': 'Germany',
 'population_millions': 83.2,
 'capital': 'Berlin',
 'region': 'Europe',
 'wto_member': True
}
````

````python
country3 = {
  'name': 'Switzerland',
  'population_millions': 8.6,
  'capital': 'Bern',
  'region': 'Europe',
  'wto_member': True,
}
````

````python
countries = [country, country2, country3]
countries
````

````python
countries[2]["name"]
````
