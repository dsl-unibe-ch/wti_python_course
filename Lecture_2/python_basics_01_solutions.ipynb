## Exercise Block 1 — Solutions

````python
# Core 1 — three variables describing a country
name = "Germany"
population_millions = 83.24
member_wto = True

# Core 2 — one sentence using all three; print() joins the pieces with spaces
print(name, "has", population_millions, "million people. Is it member of the WTO?", member_wto)
````

````python
# Core 3 — split 142 pages into full 20-page chapters plus leftovers
doc_size = 142
chapter_size = 20
num_chapters = doc_size // chapter_size   # // keeps only whole chapters
pages_left = doc_size % chapter_size      # %  keeps the remainder
print("Number of chapters:", num_chapters)
print("Pages left:", pages_left)
````

````python
# Core 4 — years elapsed since the agreement was signed (as of 2026)
year_signed = 1994
age = 2026 - year_signed
print("Age since signing:", age)
````

````python
# Stretch 1 — apply a 20% discount to the tariff rate
tariff_rate = 7.5
discount = 0.2
discounted_tariff_rate = tariff_rate * (1 - discount)
print("Discounted tariff rate:", discounted_tariff_rate)
````

````python
# Stretch 2 — "inclusive" means the bounds count, so use >= and <=
population_millions = 83.2
is_medium_sized = population_millions >= 10 and population_millions <= 100
print("Population between 10 and 100 millions:", is_medium_sized)
````

````python
# Stretch 3 — gluing text and a number with + raises TypeError; fix with str()
population_millions = 83.2
# message = "Population: " + population_millions   # would raise TypeError
message = "Population: " + str(population_millions)  # fix: convert the number to text first
print(message)
````

````python
# Challenge — peel off days, then hours, then minutes, then seconds
total_seconds = 100000

days = total_seconds // 86400        # 86400 seconds in a day
remainder = total_seconds % 86400    # seconds left after removing whole days
hours = remainder // 3600            # 3600 seconds in an hour
remainder = remainder % 3600
minutes = remainder // 60
seconds = remainder % 60

print(f"{days} day, {hours} hours, {minutes} minutes, {seconds} seconds")
````

## Exercise Block 2 — Solutions

````python
# Core 1 — chain string methods; each one returns a new string
raw = "  KENYA - Tariff Schedule 2024.CSV  "
cleaned = raw.strip().lower().replace(" ", "_")
cleaned
````

````python
# Core 2 — first, last, and length of a list
un_founders = ["United States", "Sowjet Union", "United Kingdom", "France", "China"]
print(un_founders[0])    # first item
print(un_founders[-1])   # last item
print(len(un_founders))  # how many
````

````python
# Core 3 — slice indices 3 up to (but not including) 8
symbol = "WT-DS316-AB-R"
symbol[3:8]
````

````python
# Core 4 — drop dict values into an f-string with {d['key']}
d = {"name": "Brazil", "capital": "Brasília", "population_millions": 213}
print(f"{d['name']} has a population of {d['population_millions']} million and its capital is {d['capital']}.")
````

````python
# Stretch 1 — [-3:] takes the last three; [2:5] takes the middle three
rates = [1.8, 12.0, 13.5, 4.3, 7.5, 9.0, 5.5]
print(rates[-3:])
print(rates[2:5])
````

````python
# Stretch 2 — add a new key, then update an existing one
d["is_wto_member"] = True
print(d)
d["population_millions"] = d["population_millions"] + 1
print(d)
````

````python
# Stretch 3 — a set drops duplicates, so its length = number of distinct rates
distinct_rates = set(rates)
print(len(distinct_rates))
````

````python
# Challenge 1 — slice the number out as text, then convert to float
raw = "population: 55.1 million"
str_55 = raw[12:17]        # "55.1 " (float() ignores the trailing space)
float_55 = float(str_55)
print(float_55)
print(type(float_55))
````

````python
# Challenge 2 — build a list of dicts from parallel lists, no loop
names    = ["Switzerland", "Kenya", "Brazil"]
capitals = ["Bern", "Nairobi", "Brasília"]
tariffs  = [1.8, 12.0, 13.5]

records = [
    {"name": names[0], "capital": capitals[0], "avg_tariff": tariffs[0]},
    {"name": names[1], "capital": capitals[1], "avg_tariff": tariffs[1]},
    {"name": names[2], "capital": capitals[2], "avg_tariff": tariffs[2]},
]

print(records[1]["capital"])   # capital of the second record -> Nairobi
````
