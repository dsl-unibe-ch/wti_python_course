## 1 — Python as a calculator

A code cell in a notebook evaluates whatever expression you write and shows
the result of the **last line** underneath it. Python supports the everyday
arithmetic operators, plus two you may not have met before: floor division
(`//`) and modulo (`%`).

| Operator | Name | Example | Result | Notes |
|---|---|---|---|---|
| `+` | Addition | `5 + 3` | `8` | |
| `-` | Subtraction | `5 - 3` | `2` | |
| `*` | Multiplication | `5 * 3` | `15` | |
| `/` | True division | `10 / 5` | `2.0` | **Always** returns a `float`, even when it divides evenly |
| `//` | Floor division | `12 // 5` | `2` | Divides, then throws away the remainder |
| `%` | Modulo | `12 % 5` | `2` | The remainder itself |

**Why `//` and `%` matter:** together they let you split a quantity into
"whole units" and "leftover" — e.g. pages into chapters and leftover pages,
or seconds into minutes and leftover seconds. That combination shows up
repeatedly in the exercises.

A common check: `(a // b) * b + (a % b)` always equals `a`.

## 2 — `print()` and variables

`print()` displays text (or values) on screen. Anything the notebook doesn't
explicitly print but is the last line of a cell gets shown automatically —
but inside a script, or anywhere but the last line, you need `print()` to see
anything at all.

A **variable** is a name attached to a value with `=`. Read `x = 5` as
"attach the label `x` to the value `5`," not as a mathematical equation.
`=` always works left-to-right: name on the left, value on the right.

### Variable naming rules

| Rule | Valid | Invalid | Why |
|---|---|---|---|
| No spaces | `country_of_interest` | `country of interest` | Python reads each word as separate code |
| Can't start with a digit | `country_2` | `2nd_country` | Python would try to parse a number first |
| Only letters, digits, underscores | `my_var_1` | `&country` | `&` is not a legal name character |
| Case-sensitive | `Country` ≠ `country` | — | These are two different variables |

### Dynamic typing

Python doesn't lock a variable to one type. Reassigning `my_variable` from
`25` (an `int`) to `"12"` (a `str`) is completely legal — the label just
moves to a new value of a different kind. This is different from languages
like Java or C, where a variable's type is fixed when declared.

### `+` behaves differently depending on the type

| Left | Right | `+` means | Example | Result |
|---|---|---|---|---|
| `str` | `str` | Concatenation (glue together) | `"12" + "34"` | `"1234"` |
| `int`/`float` | `int`/`float` | Arithmetic addition | `12 + 34` | `46` |
| `str` | `int`/`float` | **Not allowed** | `"12" + 34` | `TypeError` |

Mixing a string and a number with `+` is one of the most common beginner
errors — Python refuses rather than guessing what you meant.

## 3 — The four everyday types

Python infers a value's **type** from how it's written. `type(x)` tells you
what it inferred.

| Type | Example | Description |
|---|---|---|
| `int` | `316` | Whole number, no decimal point |
| `float` | `316.0` | Decimal number |
| `str` | `"Hello World"` | Text, in quotes |
| `bool` | `True` / `False` | Yes/no value |

### Converting between types

| Function | Purpose | Example | Result | Note |
|---|---|---|---|---|
| `str(x)` | number → text | `str(316)` | `"316"` | Needed to glue a number into a string with `+` |
| `int(x)` | text/float → whole number | `int(5.9)` | `5` | **Truncates** — chops off the decimal, does not round |
| `float(x)` | text → decimal number | `float("3.14")` | `3.14` | |
| `round(x)` | float → nearest whole number | `round(5.9)` | `6` | **Rounds** to the nearest integer |

`int()` and `round()` give different answers on the same input — `int(5.9)`
is `5`, `round(5.9)` is `6`. Neither one changes the original value; both
return a brand-new value that you must assign somewhere if you want to keep it.

### Comparison operators

Comparisons always produce a `bool`. Note the **double** `==` for equality —
a single `=` assigns instead, which is the single most common beginner typo.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5.5 == 10` | `False` |
| `!=` | Not equal to | `5.5 != 10` | `True` |
| `>` | Greater than | `5.5 > 10` | `False` |
| `<` | Less than | `5.5 < 10` | `True` |
| `>=` | Greater than or equal to | `10 >= 10` | `True` |
| `<=` | Less than or equal to | `10 <= 10` | `True` |

### Logical operators

Combine or invert `bool` values with plain English words, not symbols.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `and` | True only if **both** sides are true | `5 > 10 and 5 > 1` | `False` |
| `or` | True if **at least one** side is true | `5 > 10 or 5 > 1` | `True` |
| `not` | Flips a `bool` | `not (5 > 10)` | `True` |

Chaining a range check like `10 <= x <= 100` needs `and`:
`x >= 10 and x <= 100`.

## 4 — String manipulations

Strings are **immutable** — no method changes the original string in place.
Every string method returns a **new** string; you must assign it to a
variable (even the same one) to keep it.

### Common string methods

| Method | Purpose | Example | Result |
|---|---|---|---|
| `.strip()` | Remove leading/trailing whitespace | `"  hi  ".strip()` | `"hi"` |
| `.lower()` | Lowercase everything | `"HI".lower()` | `"hi"` |
| `.replace(a, b)` | Replace every occurrence of `a` with `b` | `"a-b".replace("-", "_")` | `"a_b"` |
| `.endswith(x)` | Does the string end with `x`? | `"file.csv".endswith(".csv")` | `True` |
| `.startswith(x)` | Does the string start with `x`? | `"WT-123".startswith("WT-")` | `True` |
| `len(s)` | Number of characters | `len("hi")` | `2` |

Methods **chain**: `raw.strip().lower().replace(" ", "_")` runs left to
right, each step working on the result of the one before it.

### Indexing and slicing

Positions start at **0**. A slice `[a:b]` runs from `a` up to *but not
including* `b` (half-open). `-1` refers to the last character, `-2` the
second-to-last, and so on.

| Expression | Meaning | On `"WT-DS316-AB-R"` | Result |
|---|---|---|---|
| `s[0]` | Character at position 0 | `s[0]` | `"W"` |
| `s[5]` | Character at position 5 | `s[5]` | `"S"` |
| `s[3:8]` | From index 3 up to (not incl.) 8 | `s[3:8]` | `"DS316"` |
| `s[-1]` | Last character | `s[-1]` | `"R"` |
| `s[:n]` | From the start up to `n` | | "first `n` characters" |
| `s[n:]` | From `n` to the end | | "everything from `n` onward" |

Because strings are immutable, `s[3] = "X"` raises a `TypeError`. The
workaround is to build a new string, typically with `.replace()`.

## 5 — f-strings

An f-string embeds variables directly inside text: put `f` before the
opening quote, and wrap variable names in `{ }`.

| Approach | Example | Verdict |
|---|---|---|
| `print()` with commas | `print("In", year, ",", country, "...")` | Works, but extra spaces are hard to control |
| Manual `+` concatenation | `"In " + str(year) + ", " + country + "..."` | Works, but needs `str()` everywhere and is easy to mistype |
| f-string | `f"In {year}, {country} ..."` | Cleanest — no manual conversion needed |

f-strings also **format numbers** — a colon inside the braces introduces a
format spec, e.g. fixed decimal places:

| Spec | Meaning | Example | Result |
|---|---|---|---|
| `{x}` | Value as-is | `f"{5.12986}"` | `"5.12986"` |
| `{x:.2f}` | Fixed to 2 decimal places | `f"{5.12986:.2f}"` | `"5.13"` |
| `{x:.4f}` | Fixed to 4 decimal places | `f"{5.12986:.4f}"` | `"5.1299"` |

## 6 — Lists

A **list** is an ordered, changeable collection written with square
brackets: `[10, 21, 0.512, "Hello"]`. It can mix types, and `len()` counts
its items.

Indexing and slicing on lists work exactly like on strings: position 0 is
first, `[a:b]` is half-open, and leaving a side blank means "to the end" or
"from the start."

### List methods (these change the list in place)

| Method | Purpose | Example | Effect |
|---|---|---|---|
| `.append(x)` | Add `x` to the end | `lst.append("Switzerland")` | Grows the list by one, at the end |
| `.pop(i)` | Remove the item **at index** `i` | `lst.pop(3)` | Removes and returns that item |
| `lst[i] = x` | Replace the item at index `i` | `lst[1] = "France"` | Overwrites in place |
| `.remove(x)` | Remove the **first item equal to** `x` | `lst.remove("France")` | Removes by value, not position |
| `.insert(i, x)` | Insert `x` **at position** `i` | `lst.insert(0, "Germany")` | Shifts everything after it along |

Unlike strings, lists are **mutable** — these methods really do change the
original list, no reassignment needed.

## 7 — Sets

A **set** is an unordered collection of **unique** values, built with
`set(...)`. Passing a list into `set()` drops every duplicate, which makes
sets the natural tool for "how many distinct values are there?" questions.

| Feature | List | Set |
|---|---|---|
| Order | Preserved | Not guaranteed |
| Duplicates | Allowed | Automatically removed |
| Indexing (`x[0]`) | Yes | No |
| Typical use | Ordered records | Counting distinct values |

## 8 — Dictionaries

A **dictionary** looks values up by **key** instead of position, written as
`{key: value, ...}`.

| Operation | Syntax | Example |
|---|---|---|
| Create | `{"k": v, ...}` | `{"name": "Kenya", "population_millions": 55.1}` |
| Read a value | `d["key"]` | `d["name"]` → `"Kenya"` |
| Add / overwrite a key | `d["key"] = value` | `d["capital"] = "Nairobi"` |
| Add several keys at once | `d.update({...})` | `d.update({"region": "Africa", "wto_member": True})` |

### The shape real data usually takes

A **list of dictionaries** — one dict per record — is the standard way to
represent a small table of data in plain Python:

```
countries = [country1, country2, country3]
```

To reach a field, you index **twice**: first the record (by position),
then the key (by name) — e.g. `countries[2]["name"]`.

### Containers at a glance

| Container | Ordered? | Changeable? | Duplicates? | Looked up by |
|---|---|---|---|---|
| `str` | Yes | No (immutable) | Yes | Position (index/slice) |
| `list` | Yes | Yes | Yes | Position (index/slice) |
| `set` | No | Yes | No (auto-removed) | Membership only (`in`) |
| `dict` | Yes (insertion order) | Yes | Keys must be unique | Key name |
