## 1 — Decisions with `if`

`if` runs a block only when its condition is true. The block is whatever
is **indented** below it — Python uses indentation, not braces, and four
spaces is the convention. `elif` and `else` cover the other cases; the
first matching branch wins.

```python
tariff_rate = 5.5

if tariff_rate > 10:
    category = "high"
elif tariff_rate > 5:
    category = "moderate"
else:
    category = "low"

print(category)
```

A playful example of chained conditions: rock-paper-scissors against
a random computer choice. Run it a few times and watch the outcome change.

```python
import random
choices = ["rock", "paper", "scissors"]
computer_choice = random.choice(choices)
computer_choice
```

```python
choices = ["rock", "paper", "scissors"]
computer_choice = random.choice(choices)

your_choice = "rock"

if your_choice == computer_choice:
    print(f"Both chose {your_choice}. It's a tie!")
elif your_choice == "scissors" and computer_choice == "paper":
    print("You chose scissors and computer chose paper. You win!")
elif your_choice == "paper" and computer_choice == "rock":
    print("You chose paper and computer chose rock. You win!")
elif your_choice == "rock" and computer_choice == "scissors":
    print("You chose rock and computer chose scissors. You win!")
else:
    print(f"You chose {your_choice} and computer chose {computer_choice}. You lose!")
```

## 2 — `for` loops

A `for` loop walks through a collection one item at a time. The loop
variable takes each value in turn; the indented body runs once per item.

```python
countries = ["Switzerland", "USA", "Brazil", "Kenya"]

for country in countries:
    print(f"Country: {country}")
```

A common pattern: keep a running **counter** that you bump inside the
loop. This "accumulate as you go" idea returns when we organise files.

```python
countries = ["Switzerland", "USA", "Brazil", "Kenya"]

counter = 0
for country in countries:
    counter = counter + 1
    print(f"{counter} -> {country}")
```

`range(0, 5)` produces `0, 1, 2, 3, 4` — half-open, like slicing.

```python
for i in range(0, 5):
    print(i)
```

When you want both the position and the item, `enumerate` gives you
both at once — cleaner than a hand-rolled counter.

```python
countries = ["Switzerland", "USA", "Brazil", "Uganda", "Kenya"]

for index, country in enumerate(countries):
    print(f"{index}: {country}")
```

Combine `for` with `if` to act only on some items. Two useful
keywords: `break` stops the loop entirely at the first hit; `continue`
skips the rest of the current item and moves on.

```python
for index, country in enumerate(countries):
    if country.startswith("U"):
        print(f"Found a country starting with 'U' in the list of countries at index {index}")
```

```python
for index, country in enumerate(countries):
    if country.startswith("U"):
        print(f"Found a country starting with 'U' in the list of countries at index {index}")
        break
```

```python
for index, country in enumerate(countries):
    if not country.startswith("U"):
        continue
    print(f"Processing country: {country}")
```

Looping over a dictionary with `.items()` gives you key and value together.

```python
country = {"name": "Kenya", "population_millions": 55.1}

for key, value in country.items():
    print(f"{key} -> {value}")
```

The pattern that underlies the whole file organiser: **loop, decide,
collect into a new list**. Here we gather high-tariff country names. Later
we'll do the same with files instead of countries.

```python
countries = [
    {"name": "Switzerland", "avg_tariff": 1.8},
    {"name": "Kenya",       "avg_tariff": 12.0},
    {"name": "Brazil",      "avg_tariff": 13.5},
    {"name": "Japan",       "avg_tariff": 4.3},
]

high_tariff_countries = []

for c in countries:
    if c["avg_tariff"] > 10:
        high_tariff_countries.append(c["name"])

print(high_tariff_countries)
```

## 3 — `while` loops

A `while` loop keeps going as long as its condition stays true. You must
change something inside the loop, or it runs forever — if that happens,
use the stop button in the toolbar.

```python
counter = 10
while counter > 0:
    print(counter)
    counter = counter - 1
print("Liftoff!")
```

`while` shines when you don't know in advance how many rounds you
need — like rolling a die until you get a 6. Run it a few times; the
number of rolls varies.

```python
# import random
dice_roll = random.randint(1, 6)
dice_roll
```

```python
dice = 1
rolls = 0

while dice != 6:
    dice = random.randint(1, 6)
    rolls = rolls + 1
    print(f"Rolled: {dice}")

print(f"You needed {rolls} rolls to get a 6.")
```

## 4 — Functions

A **function** is a named piece of code you define once with `def` and
call as often as you like. Start simple: no inputs, just an action.

```python
def greet():
    print(f"Welcome to the course.")
```

```python
greet()
```

Add a **parameter** so the function can vary its behaviour per call.

```python
def greet(name):
    print(f"Welcome to the course, {name}.")

greet("Matteo")
greet("Ana")
```

```python
def greet(name="Ana"):
    print(f"Welcome to the course, {name}.")

greet()
```

The key idea: `return` hands a value back to the caller, so you can
use the result elsewhere. This is different from `print`, which only
shows something on screen. Watch what `return` gives you here.

```python
def calculate_sum(a, b):
    s = a + b
    return s

calculate_sum(3, 5)
```

Functions compose nicely. Here a `while` loop is wrapped in a
function that takes the number of sides as a parameter, then called for
several dice.

```python
import random

def roll_until_six(sides):
    dice = 1
    rolls = 0
    while dice != 6:
        dice = random.randint(1, sides)
        rolls = rolls + 1
    return rolls
```

```python
dices = {"Cube": 6, "Octahedron": 8, "Dodecahedron": 10, "Icosahedron": 20}

for name, sides in dices.items():
    print(f"{name} needed {roll_until_six(sides)} rolls to get a 6.")
```

Functions calling functions — small pieces, each doing one job,
combined. This is exactly the shape of the file organiser we'll build
shortly.

```python
def convert_to_chf(amount, rate):
    return amount * rate

def print_price(amount, currency, rate):
    chf = convert_to_chf(amount, rate)
    print(f"{amount} {currency} = {chf} CHF")

print_price(100, "USD", 0.88)
print_price(50, "EUR", 0.95)
```

## 5 — Exploring a folder with `pathlib`

Everything from here uses the same patterns — loops, ifs, dicts,
functions — but points them at real files. The only new thing is the
`Path` object. `pathlib` is Python's modern, safe way to talk about
files and folders.

```python
from pathlib import Path

# Point a Path at the sample_data folder (relative to this notebook)
data_dir = Path("sample_data")
data_dir
```

```python
# Basic info
print(type(data_dir))           # Show it's a Path object
print(data_dir)                 # String representation of the path
```

```python
# Path characteristics
print(data_dir.exists())        # Does it exist?
print(data_dir.is_dir())        # Is it a directory?
print(data_dir.is_file())       # Is it a file?
print(data_dir.absolute())      # Absolute path
print(data_dir.name)            # Final component (folder/file name)
```

A `Path` is an object that knows things about a file's location.
The `/` operator joins path pieces safely — no manual slashes, no
OS-specific bugs. Each `Path` exposes handy properties: `.name`, `.stem`
(name without extension), `.suffix` (the extension), `.parent`.

```python
# Join a folder and a filename with "/" — pathlib builds a valid path on any OS
p = Path("sample_data") / "WT-DS316-AB-R.txt"
print(p)
```

```python
p = Path("sample_data") / "WT-DS316-AB-R.txt"

print("relative path :", p)
print("absolute_path :", p.absolute())
print("name      :", p.name)     # WT-DS316-AB-R.txt
print("stem      :", p.stem)     # WT-DS316-AB-R
print("suffix    :", p.suffix)   # .txt
print("parent    :", p.parent)   # sample_data
print("exists    :", p.exists())
```

```python
# iterdir() hands back a generator, not a list — it doesn't print the files yet
data_dir.iterdir()
```

```python
for rel_path in data_dir.iterdir():
    print(rel_path)
```

```python
# A generator can't be sliced — this line fails on purpose. Read the TypeError.
data_dir.iterdir()[0:5]
```

`iterdir()` yields one `Path` per entry in a folder. Combine it with
a `for` + `if` + counter — the exact pattern from the exercises — to
answer questions like "how many CSVs are here?".

```python
# The core pattern: loop, test each file's .suffix, bump a counter
count_csv = 0

for f in data_dir.iterdir():
    if f.suffix == ".csv":
        count_csv = count_csv + 1

print(f"{count_csv} CSV files")
```

Because `.name` is just a string, every string method from Part A
still works — `in`, `.startswith`, `.lower()`. This is the payoff:
yesterday's toys, today's real work.

```python
# .name is a plain string, so string tools like "in" and .lower() work on it
for f in data_dir.iterdir():
    if "asbestos" in f.name.lower():
        print(f.name)
```

```python
# Collect the officially-named filenames into a new list, then report them
official_looking = []

for f in data_dir.iterdir():
    name = f.name
    if name.startswith("WT-") or name.startswith("G_") or name.startswith("TN_"):
        official_looking.append(name)

print(len(official_looking), "files look officially named")
for n in official_looking:
    print("  ", n)
```

Counting into a **dictionary** — the pattern the final organiser
needs.

```python
# Create an empty dictionary to store file extension counts
counts = {}

# Loop through every file in the data directory
for file_path in data_dir.iterdir():
    # Extract the file extension (e.g., '.txt', '.csv')
    file_extension = file_path.suffix
    
    # Check if we've seen this extension before
    if file_extension in counts:
        # Yes - increase the counter by 1
        counts[file_extension] = counts[file_extension] + 1
    else:
        # No - this is the first time, start counting at 1
        counts[file_extension] = 1

# Print the final result showing how many files of each type
print(counts)
```

`glob` is a shortcut when you just want files matching a pattern.
`*` means "anything", so `*.csv` grabs every CSV.

```python
# glob("*.csv") grabs matching files directly; wrap in list() to count them
csvs = list(data_dir.glob("*.csv"))
print(len(csvs), "CSV files")
```

```python
# Print just the filename of each CSV
for c in csvs:
    print(c.name)
```

## 6 — Making folders, copying, and writing files

Now we change things on disk. That's fine — we only ever write into new
folders we create ourselves. We never touch `sample_data/`; if anything
breaks, just download `sample_data/` again.

`mkdir(parents=True, exist_ok=True)` means "make this folder, including any
missing parents, and don't complain if it already exists" — a safe
"ensure it exists" in one line.

```python
target = Path("organised") / "appellate_body"
target.mkdir(parents=True, exist_ok=True)

print(target.exists())
```

`shutil.copy(source, destination)` copies a file. The destination
folder must already exist — that's what the `mkdir` above was for.

```python
import shutil

source = Path("sample_data") / "WT-DS316-AB-R.txt"
destination = Path("organised") / "appellate_body" / source.name

shutil.copy(source, destination)
print(destination.exists())
```

Renaming in place with `.rename()`. `with_name(...)` gives you a new
path in the same folder but a different filename.

```python
old = destination
new = destination.with_name("renamed_example.txt")
print(old)
print(new)
```

```python
print("old exists?", old.exists())
print("new exists?", new.exists())
```

```python
old.rename(new)

print("old exists?", old.exists())
print("new exists?", new.exists())
```

`write_text(...)` writes a whole string to a file (overwriting any
existing content); `read_text()` reads it back. This is how the final
project will produce a summary file.

```python
report = Path("organised") / "summary.txt"

report.write_text(
    "Organisation summary\n"
    "--------------------\n"
    "Files handled so far: 1\n"
)

print("wrote", report)
```

```python
content = report.read_text()
print(content)
```

Everything comes together here. A function `classify` decides which
category a filename belongs to (the first matching `if` branch wins, so
order matters). Then a loop classifies every file, makes the matching
folder, and copies the file in. This little script is the whole point of
the course — read it slowly.

```python
def classify(filename):
    name = filename.lower()
    if "ab" in name and "ds" in name:
        return "appellate_body"
    elif "panel" in name:
        return "panel_report"
    elif name.endswith(".csv") or "tariff" in name:
        return "tariff_schedule"
    elif "min" in name or name.startswith("tn_"):
        return "ministerial"
    elif "agreement" in name or "understanding" in name or "gatt" in name:
        return "legal_text"
    else:
        return "unclassified"
```

```python
out_root = Path("organised_demo")

for file in data_dir.iterdir():
    file_category = classify(file.name)
    dest_folder = out_root / file_category
    dest_folder.mkdir(parents=True, exist_ok=True)
    shutil.copy(file, dest_folder / file.name)
```

```python
print("Contents of organised_demo/:")
for subfolder in out_root.iterdir():
    n = len(list(subfolder.iterdir()))
    print(f"  {subfolder.name}: {n} files")
```
