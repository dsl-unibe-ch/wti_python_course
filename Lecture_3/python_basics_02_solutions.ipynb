## Exercise Block 3 — Solutions

```python
# Core 1 — print each country's name
for c in countries:
    print(c["name"])
```

```python
# Core 2 — flag only the high-tariff countries (avg_tariff > 10)
for c in countries:
    country_tariff = c["avg_tariff"]
    if country_tariff > 10:
        print(f"{c['name']}: high tariff")
```

```python
# Core 3 — return a bool for "is this country populous?"
def is_populous(country, threshold=50):
    if country["population_millions"] > threshold:
        return True
    else:
        return False

print(is_populous(countries[0]))
print(is_populous(countries[2]))
```

```python
# Core 4 — count WTO members with a running counter
wto_member = 0
for country in countries:
    if country["wto_member"] == True:
        wto_member = wto_member + 1
print(wto_member)
```

```python
# Stretch 1 — collect only the WTO-member dicts into a new list
wto_only = []
for country in countries:
    if country["wto_member"] == True:
        wto_only.append(country)
print(len(wto_only))
```

```python
# Stretch 2 — bucket the tariff rate, then label every country
def tariff_category(rate):
    if rate <= 5:
        return "low"
    elif rate <= 10:
        return "moderate"
    elif rate <= 15:
        return "high"
    else:
        return "very high"

for c in countries:
    print(f"{c['name']}: {tariff_category(c['avg_tariff'])}")
```

```python
# Challenge — top n by tariff without sorted(): repeatedly find and remove the max
def top_n_by_tariff(countries, n=3):
    working = list(countries)   # a copy we can shrink without touching the original
    top_names = []
    for i in range(0, n):
        max_country = working[0]
        for c in working:
            if c["avg_tariff"] > max_country["avg_tariff"]:
                max_country = c
        top_names.append(max_country["name"])
        working.remove(max_country)
    return top_names

print(top_n_by_tariff(countries, n=3))
```

## Exercise Block 4 — Solutions

```python
# Core 1 — count files by looping (practising the loop, not len())
from pathlib import Path
data_dir = Path("sample_data")

counter = 0
for file in data_dir.iterdir():
    counter = counter + 1
print(f"Total number of files: {counter}.")
```

```python
# Core 2 — print the stem (name without extension) of every .txt file
for file in data_dir.iterdir():
    if file.suffix == ".txt":
        print(file.stem)
```

```python
# Core 3 — count filenames containing "ds" (lowercased for case-insensitivity)
counter = 0
for file in data_dir.iterdir():
    if "ds" in file.name.lower():
        print(file)
        counter = counter + 1
print(f"Files containing DS: {counter}.")
```

```python
# Core 4 — glob grabs only the CSVs; print each name
csvs = list(data_dir.glob("*.csv"))
for csv in csvs:
    print(csv.name)
```

```python
# Stretch 1 — count files per suffix into a dictionary
counts_by_suffix = {}
for file in data_dir.iterdir():
    suffix = file.suffix
    if suffix in counts_by_suffix:
        counts_by_suffix[suffix] = counts_by_suffix[suffix] + 1
    else:
        counts_by_suffix[suffix] = 1
print(counts_by_suffix)
```

```python
# Stretch 2 — filenames that do NOT start with WT-, G_, or TN_
probably_renamed = []
for file in data_dir.iterdir():
    if not file.name.startswith("WT-") and not file.name.startswith("G_") and not file.name.startswith("TN_"):
        probably_renamed.append(file)
print(f"There are {len(probably_renamed)} files that do no start with 'WT-', 'G_' or 'TN_':")
print("")
print("First three:")
for i in probably_renamed[:3]:
    print(i)
```

```python
# Stretch 3 — track the longest filename while looping
longest = 0
for filename in data_dir.iterdir():
    length = len(filename.name)
    if length > longest:
        longest = length
        longest_filename = filename

print(f"Longest filename: {longest_filename}, has length {longest}.")
```

```python
# Challenge — keep files matching contains AND suffix; None means "ignore this filter"
def filter_by(files, contains=None, suffix=None):
    result = []
    for f in files:
        if contains is not None and contains.lower() not in f.name.lower():
            continue
        if suffix is not None and f.suffix != suffix:
            continue
        result.append(f)
    return result

files = list(data_dir.iterdir())

print("suffix='.txt':")
for f in filter_by(files, suffix=".txt"):
    print("  ", f.name)

print("contains='ds':")
for f in filter_by(files, contains="ds"):
    print("  ", f.name)

print("both contains='ds' and suffix='.txt':")
for f in filter_by(files, contains="ds", suffix=".txt"):
    print("  ", f.name)
```

## Capstone — Solutions


```python
# Capstone Core — copy CSVs and TXTs into two subfolders, then count each
from pathlib import Path
import shutil

data_dir = Path("sample_data")
out_root = Path("my_organised")

csv_dir = out_root / "csv_files"
txt_dir = out_root / "txt_files"
csv_dir.mkdir(parents=True, exist_ok=True)
txt_dir.mkdir(parents=True, exist_ok=True)

for f in data_dir.iterdir():
    if f.suffix == ".csv":
        shutil.copy(f, csv_dir / f.name)
    elif f.suffix == ".txt":
        shutil.copy(f, txt_dir / f.name)
    # everything else is skipped

# Count what landed in each subfolder
csv_count = 0
for f in csv_dir.iterdir():
    csv_count = csv_count + 1

txt_count = 0
for f in txt_dir.iterdir():
    txt_count = txt_count + 1

print(f"csv_files: {csv_count}")
print(f"txt_files: {txt_count}")
```

```python
# Capstone Stretch — classify into six categories, count them, write a summary file
from pathlib import Path
import shutil

data_dir = Path("sample_data")
out_root = Path("my_organised")


def classify(filename):
    name = filename.lower()
    if "ab" in name and "ds" in name:
        return "appellate_body"
    elif "panel" in name:
        return "panel_report"
    elif name.endswith(".csv") or "tariff" in name:
        return "tariff_schedule"
    elif "min" in name or name.startswith("tn_"):
        return "ministerial"
    elif "agreement" in name or "understanding" in name or "gatt" in name:
        return "legal_text"
    else:
        return "unclassified"


# Sort every file into its category folder, keeping a running tally
counts = {}
for f in data_dir.iterdir():
    category = classify(f.name)
    dest_folder = out_root / category
    dest_folder.mkdir(parents=True, exist_ok=True)
    shutil.copy(f, dest_folder / f.name)
    if category in counts:
        counts[category] = counts[category] + 1
    else:
        counts[category] = 1

# Build the summary: one line per category plus a total
total = 0
lines = ["Organisation summary", "--------------------"]
for category, n in counts.items():
    lines.append(f"{category}: {n}")
    total = total + n
lines.append(f"total: {total}")

summary = out_root / "summary.txt"
summary.write_text("\n".join(lines) + "\n")

# Read it back and show it
print(summary.read_text())
```

```python
# Capstone Challenge — wrap everything in organise(), and don't overwrite:
# if the destination already exists, copy to a "_dup" name instead.
from pathlib import Path
import shutil

# Reuses classify() defined in the Stretch solution above.


def organise(source_folder, target_folder):
    source = Path(source_folder)
    target = Path(target_folder)
    counts = {}
    dups = []

    for f in source.iterdir():
        category = classify(f.name)
        dest_folder = target / category
        dest_folder.mkdir(parents=True, exist_ok=True)

        dest = dest_folder / f.name
        # If that file is already there, build a "_dup" name instead of overwriting
        if dest.exists():
            dest = dest_folder / (f.stem + "_dup" + f.suffix)
            dups.append(dest.name)

        shutil.copy(f, dest)
        if category in counts:
            counts[category] = counts[category] + 1
        else:
            counts[category] = 1

    # Write the summary, now with a Duplicates section
    total = 0
    lines = ["Organisation summary", "--------------------"]
    for category, n in counts.items():
        lines.append(f"{category}: {n}")
        total = total + n
    lines.append(f"total: {total}")
    lines.append("")
    lines.append("Duplicates:")
    for d in dups:
        lines.append(f"  {d}")

    summary = target / "summary.txt"
    summary.write_text("\n".join(lines) + "\n")
    return counts


# First run: clean copies, no collisions yet
counts1 = organise("sample_data", "my_organised_v2")
print("first run :", counts1)

# Second run: every file already exists, so each becomes a _dup copy
counts2 = organise("sample_data", "my_organised_v2")
print("second run:", counts2)

print()
print((Path("my_organised_v2") / "summary.txt").read_text())
```
