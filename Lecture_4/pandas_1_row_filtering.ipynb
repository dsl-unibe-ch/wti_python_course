# Lecture 4 supplement: Filtering rows

Filtering means selecting the rows we want.

We will select rows using **export values**, **index labels**, and **positions**.

## 1. A small practice table

Run the cells from top to bottom. This notebook works on its own.

We use ten rows so we can try all the index examples below. Export values are in millions of US dollars. **All values are invented for teaching.**

```python
import pandas as pd

my_df = pd.DataFrame({
    "country": ["Switzerland", "Germany", "Chile", "France", "Italy",
                "Spain", "Canada", "Japan", "Brazil", "Australia"],
    "export_millions_usd": [200, 400, 100, 500, 200, 800, 150, 900, 300, 600]
})

my_df
```

## 2. Filter by an export value

Select rows where exports are **exactly 200**. Use `==` to check equality:

```python
my_df[my_df["export_millions_usd"] == 200]
```

Select rows where exports are **greater than 200**. Rows with exactly 200 are excluded:

```python
my_df[my_df["export_millions_usd"] > 200]
```

## 3. Keep the filtered table for later

The examples above display selected rows. They do not change `my_df`.

To reuse a result later, store it in a new DataFrame using `=`:

```python
exports_above_200 = my_df[my_df["export_millions_usd"] > 200]
exports_above_200
```

In later cells, use the new name to work with that result:

```python
exports_above_200["country"]
```

`my_df` still contains all ten rows. Storing a result in a variable keeps it available while Python is running; it does not save a file.

## 4. Select rows by index label

The numbers on the left of our original table are its **index labels**, from `0` to `9`.

Use `.loc` to select by label. Select the row with index label `5`:

```python
my_df.loc[5]
```

Select rows with index labels **greater than 5** (labels `6`, `7`, `8`, and `9`):

```python
my_df.loc[my_df.index > 5]
```

Select rows with index labels `2`, `5`, and `8`:

```python
my_df.loc[[2, 5, 8]]
```

## 5. Select rows by position

Use `.iloc` to select by **position**, counting from `0`.

Like list slicing, `2:6` includes position `2` and stops **before** position `6`. This selects the third through sixth rows:

```python
my_df.iloc[2:6]
```

**Remember:** `.loc` uses labels; `.iloc` uses positions. In our original table, the labels and positions happen to match. After filtering, the original labels stay attached to the rows, so they may no longer match the positions.

Any row selection can be kept for later with `new_df = ...`, just as we did above.
