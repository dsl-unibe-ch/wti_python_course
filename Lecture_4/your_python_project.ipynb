## Addresses on our computer

#### > An **absolute path** gives a full address on one computer:

```text
/Users/Alex/Documents/wti_python_course/Lecture_4/tables/trade_example.csv
```

On Windows it might look like:

```text
C:\Users\Alex\Documents\wti_python_course\Lecture_4\tables\trade_example.csv
```

#### > A **relative path** gives directions from a starting location:

```text
tables/trade_example.csv
```

This means “from here, enter `tables`, then find `trade_example.csv`”. It works if **here is `Lecture_4`**. From the project root, the directions would be `Lecture_4/tables/trade_example.csv`.

### Back to the analogy: building on another planet

Imagine you see a beautiful house on someone else's planet and want to build the same house on your own. The address of their construction site will not help you find yours, because it points to a location on **their planet**.

Instead, they can give you directions relative to the construction site: “Starting at the entrance, you will find the materials in the storage room on the right.” You can follow those directions on your own planet, **as long as your site has the same layout and you start from the same place**.

Relative paths work in the same way. Your colleague's project may have a different absolute address on their computer, but `tables/trade_example.csv` will still lead to the right file if they have the same folder structure and start in the same working directory.

### Now switch to this notebook's Python code cells

A notebook brings together **explanations**, **code**, and **results**.

The **kernel** is the running Python that carries out our instructions. It remembers the names and values we create until we restart it.

Let's first check which Python is doing our work, then check its working directory.

````python
import sys
print(sys.executable)
````

**Checkpoint:** the path above includes **this project's `.venv`**. A kernel label alone is not enough to prove that. If the path points elsewhere, ask the instructor to help select the project environment before continuing.

Next ask where this kernel is standing. `Path` is a tool from Python's built-in `pathlib` library. `cwd` stands for **current working directory**. There is no extra library to install for this.

````python
from pathlib import Path
print(Path.cwd())
````

**Checkpoint:** the working directory should end in `wti_python_course/Lecture_4`.

We started Jupyter from the project root, then double-clicked `Lecture_4` and opened this notebook inside it. Jupyter normally starts the notebook's **kernel** in the folder containing the notebook, so Python is now working in `Lecture_4`.

Our **terminal** is still at the project root. Opening folders in Jupyter does not change the terminal's location.

Think of `Path.cwd()` as asking our worker: **“Where are you standing?”** Its answer tells us the starting point Python uses for relative paths.

If your result points somewhere else, compare it with the instructor before continuing. Different settings or a previously used kernel can give Python a different starting location.

## 6. Give directions to a table
**Where to type: Python code cells in this notebook.** We have a tiny practice file waiting for us:

```text
wti_python_course/                 ← project root

├── .venv/                        ← Python and installed tools

    ├── your_python_project.ipynb

        ├── trade_2022.xlsx

```

The CSV contains **invented trade figures for practice**, not observed statistics. Today we only find the file. Reading it as a table comes later.

**Predict before running:** starting in `Lecture_4`, which relative path reaches the file?

- A. `tables/trade_example.csv`
- B. `Lecture_4/tables/trade_example.csv`

Point to each folder you would pass through.

````python
# Store directions to the file. This does not open or read it.
table_path = Path("tables") / "trade_example.csv"
print(table_path)
````

Here `/` joins pieces of a path. We give the directions a name, `table_path`, so we can use them again. Now check whether they lead to a file.

````python
print(table_path.is_file())
````

**Expected:** `True`. This means Python found a file at those directions. If you get `False`, look at `Path.cwd()` and the folder tree together. Check the starting point, the spelling, and whether the data file was copied with the project.

````python
print(table_path)
````
